Correos electrónicos, publicaciones en redes sociales, reseñas de clientes, artículos de prensa, documentos legales, mensajes de texto, transcripciones de llamadas y conversaciones en plataformas digitales son solo algunos ejemplos de información que se genera principalmente a través del lenguaje humano. Como consecuencia, las organizaciones suelen disponer de grandes volúmenes de datos que contienen conocimiento potencialmente útil, pero que no pueden ser utilizados directamente por los modelos analíticos tradicionales dada su estructura.

Esta situación representa uno de los mayores desafíos de la analítica moderna. A diferencia de los datos numéricos, el lenguaje humano posee características complejas como contexto, ambigüedad, ironía, sinonimia y múltiples interpretaciones posibles. Las personas somos capaces de comprender estas sutilezas de manera casi natural gracias a nuestra capacidad de razonar, abstraer conceptos y relacionar ideas con experiencias previas. Sin embargo, para una computadora, un texto no es más que una secuencia de caracteres que carece de significado intrínseco. Por esta razón, resulta necesario desarrollar mecanismos que permitan transformar el lenguaje humano en representaciones interpretables por los algoritmos.

El **Procesamiento de Lenaguaje Natural (PLN)** es un campo que surge con el objetivo de abordar este problema, combinando conocimientos de lingüística, estadística, aprendizaje automático e inteligencia artificial para permitir que las computadoras analicen, interpreten y extraigan información útil a partir de textos y otros tipos de comunicación humana. Gracias a los avances en esta área, actualmente es posible desarrollar aplicaciones capaces de clasificar documentos, detectar sentimientos, resumir textos, traducir idiomas, responder preguntas e incluso mantener conversaciones relativamente naturales con las personas.

Desde la perspectiva de la ciencia de datos que estudiamos, una de las principales aplicaciones del **PLN** consiste en convertir información textual en variables numéricas que puedan incorporarse a modelos predictivos. El desafío aquí radica en representar adecuadamente el contenido de un texto preservando la mayor cantidad posible de información útil para el problema que se desea resolver. Para ello, en este caso estudiaremos tres métodos ampliamente utilizados en el **PLN**:

* *Codificación por frecuencia contextual*: transforma los textos en representaciones numéricas a partir de la cantidad de veces que aparecen determinadas palabras o términos dentro de los documentos.

* *Codificación por relevancia contextual*: busca identificar qué palabras aportan realmente información diferenciadora dentro de un conjunto de textos, asignando mayor importancia a aquellos términos que ayudan a distinguir unos documentos de otros.

* *Codificación por interpretación semántica*: utiliza representaciones más avanzadas capaces de capturar relaciones de significado entre palabras y conceptos, permitiendo que los algoritmos trabajen con una noción más cercana al contexto y la semántica del lenguaje humano.

Adicionalmente, aprenderemos diversas técnicas de preparación de textos orientadas a reducir la complejidad de la información antes de su análisis. Estas técnicas permiten limpiar, normalizar y simplificar los textos eliminando elementos poco relevantes, reduciendo ruido y mejorando la calidad de las representaciones que posteriormente serán utilizadas por los modelos predictivos. En la práctica, una adecuada preparación del texto suele tener un impacto tan importante como la elección del algoritmo de aprendizaje computacional.

Finalmente, incorporaremos un nuevo algoritmo dentro de nuestro conjunto de herramientas predictivas: las **Máquinas de Vectores de Soporte (SVM)**. Este algoritmo es uno de los más importantes dentro del aprendizaje supervisado y ha demostrado históricamente un excelente desempeño en problemas relacionados con clasificación de textos. Gracias a su capacidad para encontrar fronteras de separación eficientes entre diferentes categorías, el **SVM** se han convertido en una herramienta ampliamente utilizada en aplicaciones como detección de spam, clasificación temática de documentos y análisis de opiniones.

A lo largo de este caso práctico aprenderemos cómo transformar textos en información utilizable por modelos predictivos, cómo reducir la complejidad inherente al lenguaje humano y cómo aplicar estas representaciones dentro de algoritmos de aprendizaje automático. De esta manera, ampliaremos nuestro alcance más allá de los datos tabulares tradicionales y daremos un primer paso hacia el análisis de una de las fuentes de información más abundantes e importantes del mundo moderno: el lenguaje humano.

## Preparación de datos

Lleva a cabo tu plan de acción para limpiar los textos de los comentarios. Como punto de partida, extrae en un nuevo dataframe llamado `df_corpus` las columnas class y content. 

````python
df_corpus = df_comments[["class", "content"]]
df_corpus.info()
````

En este dataset crea una nueva columna llamada `content_clean` donde cambies todos los textos para que estén en minúscula.

````python
df_corpus["content_clean"] = df_corpus["content"].str.lower()
df_corpus.sample(30, random_state = 123)
````

Sustituye en esta nueva columna los textos tipo *html* de acuerdo a la siguiente tabla de expresiones regulares:

| Expresión regular | Definición | Caracter por el cual sustituir |
| ---- | ---- | ---- |
| `<br\s/>` | Código *html* para salto de linea del tipo `<br/>`. | `" "` (Espacio) | 
| `<.*?>` | Inicio y cierre de bloque de código *html*. | `" "` (Espacio) |
| `\&#39;` | Representación en código *html* de comilla simple. | `"\'"` (Comilla simple) |
| `\&lt;` | Representación en código *html* de "<". | `"<"` (Menor que) |
| `\&#*\w*[0-9]*;*` | Otras representaciones de caracteres especiales que no son letras. | `" "` (Espacio) |

Con este propósito, utiliza para cada una la función `re.sub` dentro de métodos `apply`.

````python
# Sustituir codigo html de saltos de linea
df_corpus["content_clean"] = df_corpus["content_clean"].apply(lambda x: re.sub(r"<br\s/>"," ",x))

# Sustituir inicios y cierres de bloques de codigo html
df_corpus["content_clean"] = df_corpus["content_clean"].apply(lambda x: re.sub(r"<.*?>"," ",x))

# Sustituir codigo html de comilla simple
df_corpus["content_clean"] = df_corpus["content_clean"].apply(lambda x: re.sub(r"\&#39;","\'",x))

# Sustituir codigo html de "menor que"
df_corpus["content_clean"] = df_corpus["content_clean"].apply(lambda x: re.sub(r"\&lt;","<",x))

# Sustituir codigos html de otros caracteres especiales
df_corpus["content_clean"] = df_corpus["content_clean"].apply(lambda x: re.sub(r"\&#*\w*[0-9]*;*"," ",x))

df_corpus.sample(30, random_state = 123)
````

Sustituye los términos vinculados a emojis (i.e. ":d", ":)", ":/") por un espacio mediante las expresiones regulares `:[\w()]` y `<3`. 

````python
df_corpus["content_clean"] = df_corpus["content_clean"].apply(lambda x: re.sub(r":[\w()/]"," ",x))
df_corpus["content_clean"] = df_corpus["content_clean"].apply(lambda x: re.sub(r"<3"," ",x))

df_corpus.sample(30, random_state = 123)
````

Sustituye ahora todos los caracteres que no sean de tipo letra, espacios o la comilla simple por un espacio. Esto es, cambia aquellos caracteres que sigan la expresión regular `[^\w\s\']`.

````python
df_corpus["content_clean"] = df_corpus["content_clean"].apply(lambda x: re.sub(r"[^\w\s\']"," ",x))

df_corpus.sample(30, random_state = 123)
````

En estos textos además existen palabras mal escritas, otros códigos extraños que no son *html* y jergas particulares del idioma inglés. No nos interesan estos casos por lo que vamos a mantener solamente aquellos términos reconocidos usualmente en el idioma inglés. Para limpiar estos casos, utiliza el diccionario `web2` como argumento de la función `get_english_words_set` de la libreria **english_words**, y únelo con el diccionario `en_core_web_sm` que provee **spacy**. Este último debe instalarse previamente por la terminal con el siguiente comando:

```ps
python -m spacy download en_core_web_sm
```

Asegúrate que este diccionario unificado tenga todas sus palabras en minúsucla, e incorpora además los nombres de los artistas que existen en el dataset y los términos clave de sus canciones pues son términos propios de nuestro contexto.

````python
# Extraer y unir diccionarios de terminos en ingles
dicc_gews = get_english_words_set(["web2"], lower = True)
dicc_spacy = spacy.load("en_core_web_sm")
dicc_spacy = set(dicc_spacy.vocab.strings)
dicc_en = dicc_gews.union(dicc_spacy)

# Poner todas las palabras del diccionario en minuscula
dicc_en = {x.lower() for x in dicc_en}

# Añadir nombres de artistas y cancionaes
dicc_en.update({
    "shakira", "eminem", "katy", "perry", "lmfao", "psy", "rihanna",
    "party", "rock", "waka", "roar", "gangnam", "style", "love", "way", "lie"
})
print("DICCIONARIO DE TERMINOS EN INGLES:")
print(f"Cantidad de términos: {len(dicc_en):,.0f}")
print(random.sample(list(dicc_en), 10))
````

````python
# Generar funcion para limpiar textos con palabras propias del diccionario 
def fun_palabras_reales (texto):
    palabras = texto.split()
    palabras_reales = []
    for p in palabras:
        try:
            # Incorporar numeros
            palabras_reales.append(str(int(p)))
        except:
            # Incorporar palabras
            if p in dicc_en:
                palabras_reales.append(p)
    return " ".join(palabras_reales)

# Aplicar funcion en textos
df_corpus["content_clean"] = df_corpus["content_clean"].apply(fun_palabras_reales)

df_corpus.sample(30, random_state = 123)
````

Dado que hemos quitado diversos caracteres a cambio de espacios, conviene además:

* Eliminar los espacios sobrantes al inicio y al final de texto.
* Eliminar los espacios entre palabras excesivos con la expresión regular `\s+`.

Has estos ajustes en el orden indicado.

````python
# Eliminar espacios sobrantes al inicio y al final del texto
df_corpus["content_clean"] = df_corpus["content_clean"].str.strip()

# Eliminar espacios excesivos
df_corpus["content_clean"] = df_corpus["content_clean"].apply(lambda x: re.sub(r"\s+"," ",x))

df_corpus.sample(30, random_state = 123)
````

Ahora bien, en una nueva columna llamada `content_lemma` estandariza finalmente los textos a través de una **lematización**. Este proceso consiste en eliminar conjugaciones de verbos, y flexiones o declinaciones de sustantivos, a fin de facilitar su transformación posterior. Puedes ayudarte para esto nuevamente con el diccionario `en_core_web_sm` de la librería **spacy**.

````python
# Cargar modelo lematizador
mod_lematizador = spacy.load("en_core_web_sm")

# Lematizar textos
for i in df_corpus.index:
    texto = mod_lematizador(df_corpus.loc[i,"content_clean"])
    texto = " ".join([x.lemma_ for x in texto])
    df_corpus.loc[i,"content_lemma"] = texto.lower()

df_corpus.sample(30, random_state = 123)
````

### Codificación por frecuencia contextual

Esta codificación parte de una lógica simple (pero no por eso menos conveniente) tal que cada texto se divide en n-gramas o frases de una o más palabras. A partir de allí se hace un simple conteo de veces en que dichos términos se presentan en cada texto. Tómese por ejemplo la frase 

*"mi nombre es juan cual es tu nombre"*

La codificiación por frecuencia la transformaría en lo siguiente: 

| Término | Frecuencia |
| ---- | ---- |
| "mi" | 1 |
| "nombre" | 2 |
| "es" | 2 |
| "juan" | 1 |
| "cual" | 1 |
| "tu" | 1 |


De esta forma, se da una suerte de ponderación numérica a aquellos términos más comúnmente utilizados en el contexto; y este número puede servir como atributo para ser considerado por un algoritmo en un modelo predictivo.

Vale señalar que resulta conveniente excluir de esta codificación a las llamadas "palabras de para", puesto que las mismas no aportan ninguna información más allá de ser nexos o puentes entre ideas relevantes. Volviendo a nuestro ejemplo, si quitamos los pronombres y preposiciones la transformación quedaría así:

| Término | Frecuencia |
| ---- | ---- |
| "nombre" | 2 |
| "juan" | 1 |
| "cual" | 1 |

Notemos cómo la exclusión de estas palabras da mayor enfoque al contenido relevante del texto. ### Codificación por relevancia contextual

También llamada codificación **TF-IDF**, este procedimiento va un paso más allá respecto al mostrado anteriormente, ya que además de calcular la frecuencia de un término específico (*TF: Text Frecuency*), se incorpora un factor adicional que "castiga" el continuo surgimiento de cada término en el conjunto de textos estudiados (*IDF: Inverse of Document Frecuency*). En este sentido, una palabra que aparece siempre en distintos textos tiene menor relevancia a nivel de cada caso específico, puesto que la información nueva que aporta es menor en el contexto de todo el documento.

La forma de calcular esta relevancia $R_{n,t}$ para un término $t$, que se repite $f_{t,d}$ veces dentro de un texto $d$ que a su vez es parte de un conjunto de textos $D$, se define a partir de la siguiente fórmula:

$$ R_{t,d} = TF_{t,d} \times \log (IDF_{t,d}) = \frac{f_{t,d}}{\sum_{i\in d} f_{i,d}} \times \log \left( \frac{|D|}{|{d\in D: t\in d}|} \right) $$

Para entender mejor lo que aquí se busca, volvamos a nuestro ejemplo de la frase *"mi nombre es juan cual es tu nombre"*, con su codificación por frecuencia dada por:

| Término | Frecuencia | Frecuencia Relativa (TF) |
| ---- | ---- | ---- |
| "nombre" | 2 | 0.50 |
| "juan" | 1 | 0.25 |
| "cual" | 1 | 0.25 |

Consideremos ahora una respuesta dada por *"hola mi nombre es ana"*, que tendría su propia codificación por frecuencia:

| Término | Frecuencia | Frecuencia Relativa (TF) |
| ---- | ---- | ---- |
| "hola" | 1 | 0.33 |
| "nombre" | 1 | 0.33 |
| "ana" | 1 | 0.33 |

Para nosotros como humanos es claro que ambos textos buscan comunicar los nombres de los emisores; pero además de esto, quisiéramos saber que características adicionales destacan en ambas frases y las distinguen. Calculemos por tanto el componente **IDF** de los términos del cada texto:

| Término | Frecuencia | Frecuencia Relativa (TF) | Inversa de la Frecuencia del Documento (IDF) |
| ---- | ---- | ---- | ---- |
| "nombre" | 2 | 0.50 | 1.00 |
| "juan" | 1 | 0.25 | 2.00 |
| "cual" | 1 | 0.25 | 2.00 |

<br />

| Término | Frecuencia | Frecuencia Relativa (TF) | Inversa de la Frecuencia del Documento (IDF) |
| ---- | ---- | ---- | ---- |
| "hola" | 1 | 0.33 | 2.00 |
| "nombre" | 1 | 0.33 | 1.00 | 
| "ana" | 1 | 0.33 | 2.00 |

Entonces, aplicando la fórmula precedente se tiene que la relevancia implícita de términos sería:

| Término | Frecuencia | Frecuencia Relativa (TF) | Inversa de la Frecuencia del Documento (IDF) | Relevancia Contextual |
| ---- | ---- | ---- | ---- | ---- |
| "nombre" | 2 | 0.50 | 1.00 | 0.50 * log(1.00) = **0.00** |
| "juan" | 1 | 0.25 | 2.00 | 0.25 * log(2.00) = **0.08** |
| "cual" | 1 | 0.25 | 2.00 | 0.25 * log(2.00) = **0.08** |

<br />

| Término | Frecuencia | Frecuencia Relativa (TF) | Inversa de la Frecuencia del Documento (IDF) | Relevancia Contextual |
| ---- | ---- | ---- | ---- | ---- |
| "hola" | 2 | 0.33 | 2.00 | 0.33 * log(2.00) = **0.10** |
| "nombre" | 1 | 0.33 | 1.00 | 0.33 * log(1.00) = **0.00** |
| "ana" | 1 | 0.33 | 2.00 | 0.33 * log(2.00) = **0.10** |

Notemos que este indicador evidencia justamente estas características informativas adicionales que deseábamos conocer:

* Aparte de que "Juan" dé su nombre, está preguntando "cual" es el nombre de la otra persona.
* Aparte de responder con su nombre, "Ana" saluda a Juan con un "hola".

### Codificación por interpretación semántica

El desarrollo tecnológico de los últimos años ha permitido contar con diversos modelos capaces de procesar grandes volúmenes de datos a fin de traducir textos con una suerte de comprensión semántica del contenido de información. Estos modelos **LLM (Large Language Models)** han dado cabida al surgimiento de herramientas muy conocidas actualmente como Gemini (Google), ChatGPT/Copilot (OpenAI - Microsoft), Claude (Anthropic), entre muchas otras; las cuales son capaces de interpretar el sentido y contenido de palabras, frases o documentos, prediciendo respuestas coherentes al contexto de las mismas.   

De forma bastante resumida, a continuación puedes visualizar el proceso que utilizan estos modelos para poder realizar sus tareas:

1. Reciben como entrada una consulta a modo de palabra, frase y/o documento.

2. Transforman los términos de esta entrada en elementos conocidos como **tokens**. Para esto, el modelo cuenta con un diccionario propio predefinido que relaciona cada componente de la entrada con un identificador numérico único.

3. Utilizan estos tokens como atributos de un modelo pre-entrenado de **codificación**. Por lo general este algoritmo consiste de una red neuronal compleja llamada **transformador** que es capaz de relacionar semánticamente a cada uno de los tokens recibidos mediante pesos o coeficientes numéricos. A partir de aquí se genera como predicción un vector que contiene **marcadores semánticos** o **incrustaciones** que actúan como una suerte de representación interna del significado percibido por el modelo respecto a esos tokens y al contexto general de la entrada.

4. Utilizan estas incrustaciones como entradas para otra red neuronal de **decodificación**, que traduce estas señales en base a criterios probabilísticos para seleccionar nuevos tokens que mejor complementen la palabra, frase y/o documento recibido inicialmente. 

5. Transforman los tokens definidos en términos "humanos" utilizando nuevamente el diccionario predefinido, y dando así una respuesta a la consulta inicial.


<div class="alert alert-info">

Como puedes evidenciar, un modelo LLM cuenta básicamente con tres estructuras ya establecidas que se usan secuencialmente a lo largo del proceso: 

* Diccionario de tokens. 
* Algoritmo codificador de tokens a incrustaciones. 
* Algoritmo decodificador de incrustaciones.
* Diccionario de tokens (de nuevo).

## Creación de modelo base

El algoritmo **SVM** es un método de clasificación bastante popular debido principalmente a su eficiencia computacional y a su lógica consistente y robusta. La idea detrás del mismo es que si un conjunto de datos es en principio "separable" en grupos o categorías, entonces es posible construir una frontera que los distinga, tal que la distancia de esta con la observación más cercana de cada grupo (conocida como margen) se maximice. En las siguiente visualización se aclara de mejor manera este criterio para el caso de dos atributos $x_1$ y $x_2$.

![](https://github.com/jsaraujo5081/material_curso_data_science/blob/main/complementary_files/svm.png?raw=true)

Para efectos de su aplicación práctica, este algoritmo tiene los siguienes hiperparámetros relevantes:

* **C**: Grado de tolerancia a errores en la clasificación. Un valor menor implica que el modelo generado es más tolerante a errores, simplificando de este modo la forma que puede tener la mencionada frontera. Un valor mayor por otra parte, genera un modelo menos propenso a cometer errores, con una frontera más compleja pero potencialmente presentando sobreajuste dada su especificidad. En un algoritmo base se suele emplear un valor de $C=1$.

* **Gamma ($\gamma$)**: nivel de influencia de cada observación en la definición de la frontera. Un valor menor implica una influencia mayor de cada observación y por tanto unas fronteras más suaves, mientras que un valor mayor describe fronteras más irregulares al evidenciarse efectos locales más marcados. En un algoritmo base suele emplearse un valor de $1/(k\times V_X)$ donde $k$ es el número de atributos $V_X$ es la varianza total de totos los datos en la matriz de atributos $X$.

### Modelo SVM con codificación por frecuencia contextual

Crea, entrena y evalúa un modelo SVM usando como atributos los conteos obtenidos de la codificación por frecuencia. Utiliza la función `SVC` con los argumentos base para `gamma` y `C`.

````python
#Crear y entrenar modelo SVM
mod_svm_frec = SVC(
    gamma = 1 / (X_codfrec_train.shape[1]*np.array(X_codfrec_train).var()),
    C = 1
)
mod_svm_frec.fit(X_codfrec_train,y_train)
````

````python
# Predecir con conjunto de prueba
y_pred_frec = mod_svm_frec.predict(X_codfrec_test)

# Generar matriz de confusion
print("MATRIZ DE CONFUSION:")
print(pd.crosstab(
    ["Normal" if x == 0 else "Spam" for x in y_test], 
    ["Normal" if x == 0 else "Spam" for x in y_pred_frec], 
    margins = True, 
    rownames = ["Real"], 
    colnames = ["Predicho"]
))
print()

# Generar reporte de clasificación
print("REPORTE DE METRICAS:")
print(metrics.classification_report(y_test, y_pred_frec, digits = 4, target_names = ["Normal", "Spam"]))
````

## Optimización de modelo

En primera instancia notemos que existe desbalanceo de clases. Por tanto, ajusta el modelo anterior incorporando el argumento `class_weight = "balanced"` y mira el rendimiento alcanzado. Recuerda que este argumento aplica en aquellos algoritmos que trabajan con pesos, y justamente el **SVM** lo hace. 

````python
# Crear y entrenar modelo SVM
mod_svm_bb = SVC(
    gamma = 1 / (X_codbert_train.shape[1]*np.array(X_codbert_train).var()),
    C = 1,
    class_weight = "balanced"
)
mod_svm_bb.fit(X_codbert_train, y_train)
````

````python
# Predecir con conjunto de prueba
y_pred_bb = mod_svm_bb.predict(X_codbert_test)

# Generar matriz de confusion
print("MATRIZ DE CONFUSION:")
print(pd.crosstab(
    ["Normal" if x == 0 else "Spam" for x in y_test], 
    ["Normal" if x == 0 else "Spam" for x in y_pred_bb], 
    margins = True, 
    rownames = ["Real"], 
    colnames = ["Predicho"]
))
print()

# Generar reporte de clasificación
print("REPORTE DE METRICAS:")
print(metrics.classification_report(y_test, y_pred_bb, digits = 4, target_names = ["Normal", "Spam"]))
````

No hay mucha diferencia con respecto a la situación precedente, aunque vale destacar que la capacidad predictiva de SPAM ha aumentado. Esto es un indicio que vamos por buen camino. 

Quizás una reducción dimensional puede ayudar a incrementar la asertividad dada la existencia de multicolinealidades y redundancias de información. Para esto, transforma los datos mediante PCA con solamente 200 componentes.

````python
# Transformar atributos mediante PCA
pca = PCA(n_components = 200)
pca.fit(X_codbert_train)
X_pca_train = pca.transform(X_codbert_train)
X_pca_test = pca.transform(X_codbert_test)
print(X_pca_train.shape)
print(X_pca_test.shape)
````

Genera, entrena y evalúa un modelo basado en estos datos.

````python
# Crear y entrenar modelo SVM
mod_svm_bbp = SVC(
    gamma = 1 / (X_pca_train.shape[1]*np.array(X_pca_train).var()),
    C = 1,
    class_weight = "balanced"
)
mod_svm_bbp.fit(X_pca_train, y_train)
````

````python
# Predecir con conjunto de prueba
y_pred_bbp = mod_svm_bbp.predict(X_pca_test)

# Generar matriz de confusion
print("MATRIZ DE CONFUSION:")
print(pd.crosstab(
    ["Normal" if x == 0 else "Spam" for x in y_test], 
    ["Normal" if x == 0 else "Spam" for x in y_pred_bbp], 
    margins = True, 
    rownames = ["Real"], 
    colnames = ["Predicho"]
))
print()

# Generar reporte de clasificación
print("REPORTE DE METRICAS:")
print(metrics.classification_report(y_test, y_pred_bbp, digits = 4, target_names = ["Normal", "Spam"]))
````

Mejor aún, pero aún podemos hacer más. Optimiza el modelo alcanzado hasta ahora mediante una asignación adecuada de hiperparámetros $\gamma$ y $C$.

````python
# Definir posibles gamma y C
g_opciones = np.array(range(1,21)) / 1000
c_opciones = np.array(range(1,21))

# Modelar con distintos hiperparametros
g_vals = []
c_vals = []
acc_vals = []
for g in g_opciones:
    for c in c_opciones:
        
        g_vals.append(g)
        c_vals.append(c)

        # Crear y entrenar modelo SVM
        modelo = SVC(
            gamma = g, 
            C = c,
            class_weight = "balanced"
        )
        modelo.fit(X_pca_train, y_train)

        # Predecir con conjunto de prueba
        y_pred = modelo.predict(X_pca_test)

        # Calcular exactitud
        acc = metrics.accuracy_score(y_test, y_pred)
        acc_vals.append(acc)

# Organizar resultado en dataframe
df_resultado = pd.DataFrame(dict(
    gamma = g_vals,
    c = c_vals,
    exactitud = acc_vals
))

# Encontrar hiperparámetros optimos
hp_opt = df_resultado[df_resultado["exactitud"] == df_resultado["exactitud"].max()].iloc[0]
print(hp_opt)
````

````python
# Crear y entrenar modelo optimo
mod_optimo = SVC(
    gamma = hp_opt["gamma"], 
    C = hp_opt["c"],
    class_weight = "balanced"
)
mod_optimo.fit(X_pca_train, y_train)

# Predecir con conjunto de prueba
y_pred = mod_optimo.predict(X_pca_test)

# Generar matriz de confusion
print("MATRIZ DE CONFUSION:")
print(pd.crosstab(
    ["Normal" if x == 0 else "Spam" for x in y_test], 
    ["Normal" if x == 0 else "Spam" for x in y_pred], 
    margins = True, 
    rownames = ["Real"], 
    colnames = ["Predicho"]
))
print()

# Generar reporte de clasificación
print("REPORTE DE METRICAS:")
print(metrics.classification_report(y_test, y_pred, digits = 4, target_names = ["Normal", "Spam"]))
````
