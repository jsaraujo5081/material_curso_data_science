## Manejo de variables y almacenamiento de información

Las variables constituyen uno de los elementos más importantes dentro de la programación, ya que permiten almacenar y manipular información durante la ejecución de un programa. A través de ellas es posible representar datos numéricos, textos, valores lógicos y otros tipos de información necesarios para resolver problemas computacionales. En **Python**, las variables se caracterizan por su flexibilidad y facilidad de uso, permitiendonos asignar valores de manera dinámica y reutilizarlos posteriormente en operaciones, cálculos y procesos lógicos.

## Gestión de errores y control de excepciones

La gestión de errores corresponde al conjunto de mecanismos que permiten identificar, controlar y responder adecuadamente a situaciones inesperadas que pueden ocurrir durante la ejecución de un programa computacional. Estos errores pueden surgir por entradas inválidas, operaciones no permitidas o fallos en la lógica implementada. En programación, resulta fundamental anticipar este tipo de situaciones para evitar interrupciones inesperadas y mejorar la robustez de los análisis, facilitando así la creación de programas más seguros, comprensibles y estables.

"ZeroDivisionError" hace referencia a que en el código existe una operación con divisor de 0, lo cual **Python** no es capaz de procesar.

````python
try:
    nombre = "Aleph"
    activos = 2000
    pasivos = 2000
    patrimonio = activos - pasivos
    razon_end = pasivos / patrimonio
    print("El patrimonio de", nombre, "es de", patrimonio, "USD, y su razon de endeudamiento es de", razon_end)
except:
    print("Existe algun error") 
````

## Manipulación y procesamiento de cadenas de texto

Las cadenas de texto representan secuencias de caracteres utilizadas para almacenar y procesar información textual. Su manipulación es esencial en múltiples aplicaciones, como el análisis de datos, la validación de información, la automatización de procesos y la interacción con usuarios. El manejo de cadenas por consiguiente incluye operaciones como concatenación, búsqueda, reemplazo, extracción y transformación de texto. Gracias a las herramientas incorporadas en **Python**, es posible realizar este tipo de tareas de forma sencilla y eficiente, facilitando el procesamiento de información escrita.

### Ejercicio 1

John intentó enviar un mensaje a su amigo Max para decirle que le había enviado bitcoins de regalo. Al ejecutar el programa con dicho mensaje se generó un error:

``` py
message = 'Hey pal. I've sent you some bitcoins. Check your wallet!'
print(message)
```

```py
SyntaxError: unterminated string literal (detected at line 1)
```

"SyntaxError" hace referencia a que existe algún problema *gramatical* en el código. Recordemos que **Python** es un lenguaje, y como tal tiene una reglas de escritura específicas que se deben cumplir.

Identifica donde está generándose el error y corrígelo adicionando el caracter `\` justo antes.

````python
message = 'Hey pal. I\'ve sent you some bitcoins. Check your wallet!'
print(message)
````

En efecto, el error se estaba presentando porque dentro del mensaje existía una comilla, la cual en el lenguaje de **Python** se interpreta como la apaertura o cierre de un valor de tipo texto. Para evitar esta confusión, añadimos `\` lo cual genera una notificación a **Python** para que "escape" de su regla predefinida. Justamente por esta razón es que este tipo de expresiones en textos se conoce como *caracteres de escape*.  

<div class="alert alert-info">
<b>Consejo útil:</b> Si quieres saber más sobre otros caracteres de escape comunes puedes consultar esta <a href="https://docs.python.org/3/reference/lexical_analysis.html#escape-sequences"> documentación </a>. A medida que avancemos con el aprendizaje notarás que utilizamos otros con frecuencia. Por ahora te sugiero investigar sobre lo que hacen estos dos: <code>\n</code>, <code>\t</code>.      
</div>

La computadora de Max solamente permite imprimir 50 caracteres, ¿Se ajusta el mensaje corregido a esta limitación? Compruébalo mediante la función `len`.

````python
length = len(message)
print(f"Logitud del mensaje: {length}")
print(f"Cumple criterio de 50 caracteres: {length <= 50}")
````

Una vez corregido y enviado el mensaje. Max respondió a John, pero la misma venía con los siguientes problemas:

* Caracteres “\#” entre todas las palabras.
* Espacios tanto al principio como al final del mensaje.
* Letras en mayúsculas y minúsculas de forma aleatoria.

```py
response = "  hEY#bRO,#I#nEed#YouR#hElP#bUYiNg#bTc#tONigHt!     "
```

Ayuda a John a descifrar el mensaje de respuesta utilizando los métodos `strip`, `replace` y `lower`.

````python
# Definir el mensaje original
response = "  hEY#bRO,#I#nEed#YouR#hElP#bUYiNg#bTc#tONigHt!     "

# Eliminar los espacios al inicio y al final del texto
response = response.strip()

# Reemplazar los "#" por espacios
response = response.replace("#", " ")

# Poner todos los caracteres en minuscula
response = response.lower()

print(response)
````

<div class="alert alert-warning">

<b>Cuidado:</b> Funciones como <code>print</code> o <code>type</code> pueden aplicarse a variables de distintos tipos, sin embargo los métodos no. Esto últimos son un tipo de funciones especiales que solamente se utilizan en tipologías específicas. Por ejemplo, <code>strip</code>, <code>replace</code> o <code>lower</code> son válidos para textos pero NUNCA para flotantes o enteros. De esta explicación se desprende el hecho que su sintaxis es distinta: 

* Funciones: <code>funcion(variable, argumentos)</code> 
* Métodos:   <code>variable.metodo(argumentos)</code>.
</div>

<b>Consejo útil:</b> Seguramente supones que así como existe el método <code>lower</code> para transformar los caracteres de un texto a minúsculas, igualmente debe existir el método <code>upper</code> para convertirlo todo a mayúsuclas. ¡Estás en lo correcto! De hecho, también podemos aplicar el método <code>capitalize</code> para que las frases de un texto mantengan el formato de una oración. *To my lovely wife I leave my key in the safe. Behind the statue of Dio you have a picture of us both.*

Después incluyó instrucciones para decodificarla:

* "Recortar" es lo que más te debe gustar.
* A la longitud le restas setenta y dos, y repites con 69. ¡Qué gran pro!
* Lo que está entre treinta y tres y treinta y cinco es tu segundo ahínco.

### Ejercicio 2

John envió a Max una lista adicional de buenos libros sobre bitcoin. Es un buen amigo, así que estructuró esta información a modo de lista de **Python** e incluso añadió los autores al final de esta lista. 

``` py
books_all = [
    ["Mastering Bitcoin", "USA", 2014, "technology", "M. Antonopoulos"],
    ["The Age of Cryptocurrency", "USA", 2015, "economics", "P. Vigna, J. Casey"],
    ["Digital Gold", "USA", 2015, "finance", "N. Popper"],
    ["The Bitcoin Standard", "USA", 2018, "cryptocurrency", "S. Ammous"],
    ["Bitcoin Billionaires", "USA", 2019, "biography", "B. Mezrich"]
]
```

Pero Max notó que hay un libro con más de un autor, y quiere dividir estos autores para almacenarlos por separado en forma de lista. Busca entonces el libro con más de un autor, extráelo utilizando indexación, divide los nombres de los autores en una nueva lista con el método `split` e imprime el resultado.

````python
# Definir lista original de libros
books_all = [
    ["Mastering Bitcoin", "USA", 2014, "technology", "M. Antonopoulos"],
    ["The Age of Cryptocurrency", "USA", 2015, "economics", "P. Vigna, J. Casey"],
    ["Digital Gold", "USA", 2015, "finance", "N. Popper"],
    ["The Bitcoin Standard", "USA", 2018, "cryptocurrency", "S. Ammous"],
    ["Bitcoin Billionaires", "USA", 2019, "biography", "B. Mezrich"]
]

# Extraer autores de libro con más de un autor
authors = books_all[1][-1]

# Dividir nombres de autores en lista
authors_split = authors.split(", ")

# Imprimir el resultado
print(authors_split)
````

Modifica la lista original de forma que todos los autores aparezcan a modo de lista e imprime el resultado (Si gustas puedes prescindir de `print` para una mejor visualización del resultado).

````python
books_all[0][-1] = [books_all[0][-1]]
books_all[1][-1] = authors_split
books_all[2][-1] = [books_all[2][-1]]
books_all[3][-1] = [books_all[3][-1]]
books_all[4][-1] = [books_all[4][-1]]

books_all
````
