## Concatenación de dataframes

### Ejercicio 1

Considera los siguientes tres datasets de ejemplo:

```py
df1 = pd.DataFrame({
    "col_A": [0, 1, 2, 3],
    "col_B": ["b0","b1","b2","b3"],
    "col_C": ["c0","c1","c2","c3"],
    "col_D": [4, 15, 26, 37]
}, index = [0, 1, 2, 3])

df2 = pd.DataFrame({
    "col_1": ["val_1","val_2","val_3","val_4"],
    "col_2": [45, 32, 67, 104],
    "col_3": [-1.3, 1.45, 0.76, 0.91],
    "col_4": ["x_1","x_2","x_3","x_4"]
}, index = [1, 2, 3, 4])

df3 = pd.DataFrame({
    "col_A": [8, 9, 10],
    "col_B": ["b4","b5",np.nan],
    "col_C": ["c4","c5",np.nan]
}, index = ["f1","f2","f3"])
```

Cárgalos en tu archivo importando primero las librerías **Pandas** y **Numpy**.

````python
# Cargar librerias
import pandas as pd
import numpy as np

# Cargar datasets
df1 = pd.DataFrame({
    "col_A": [0, 1, 2, 3],
    "col_B": ["b0","b1","b2","b3"],
    "col_C": ["c0","c1","c2","c3"],
    "col_D": [4, 15, 26, 37]
}, index = [0, 1, 2, 3])

df2 = pd.DataFrame({
    "col_1": ["val_1","val_2","val_3","val_4"],
    "col_2": [45, 32, 67, 104],
    "col_3": [-1.3, 1.45, 0.76, 0.91],
    "col_4": ["x_1","x_2","x_3","x_4"]
}, index = [1, 2, 3, 4])

df3 = pd.DataFrame({
    "col_A": [8, 9, 10],
    "col_B": ["b4","b5",np.nan],
    "col_C": ["c4","c5",np.nan]
}, index = ["f1","f2","f3"])
````

Imprime los 3 datasets para visualizarlos de manera más clara.

````python
print(df1)
print()
print(df2)
print()
print(df3)
````

### Ejercicio 2

Concatena df1, df2 y df3 por filas utilizando la función `concat` y el argumento `axis = 0`.

````python
resultado = pd.concat(
    [df1, df2, df3], 
    axis = 0
)
print(resultado)
````

<div class="alert alert-info">

El argumento `axis = 0` está predefinido por defecto, por lo que si no lo especificas en tu código tu resultado será una concatenación por filas.

</div>

### Ejercicio 3

Concatena df1, df2 y df3 por columnas usando la función `concat` y su argumento `axis = 1`.

````python
resultado = pd.concat(
    [df1, df2, df3], 
    axis = 1
)
print(resultado)
````

<div class="alert alert-info">

En general la función `concat` hace una unión de tablas tipo *outer* tomando como clave los nombres de filas o columnas. En el caso de concatenación por filas (`axis = 0`) la unión considera las columnas, mientras que en la concatenación por columnas (`axis = 1`) se consideran las filas como clave.

Si deseas probar una unión tipo *inner* en este caso (que excluirá valores ausentes), puedes modificar el argumento `join`. 

</div>

## Unión de dataframes

### Ejercicio 1

Considera ahora los siguientes dataframes de ejemplo:

```py
dataset_1 = pd.DataFrame({
    "key": ["a","b","c","d"],
    "col_A": [45, 50, 32, 21],
    "col_B": [True, True, False, True],
})

dataset_2 = pd.DataFrame({
    "key": ["a","b","d"],
    "col_A": ["pera", "manzana", "mandarina"],
    "col_B": [0, 1, 0]
})
```

Cárgalos e imprímelos en notebook.

````python
# Cargar datasets
dataset_1 = pd.DataFrame({
    "key": ["a","b","c","d"],
    "col_A": [45, 50, 32, 21],
    "col_B": [True, True, False, True],
})

dataset_2 = pd.DataFrame({
    "key": ["a","b","d"],
    "col_A": ["pera", "manzana", "mandarina"],
    "col_B": [0, 1, 0]
})

# Imprimir datasets
print(dataset_1)
print()
print(dataset_2)
````

Usa el método `merge` para unir los datasets con un *inner join*, y usando la columna key como clave.

````python
resultado = dataset_1.merge(
    dataset_2,
    how = "inner",
    on = "key"
)
print(resultado)
````

Realiza la misma unión que en el caso anterior, solo que ahora especifica sufijos distintos para columnas con el argumento `suffixes`.

````python
resultado = dataset_1.merge(
    dataset_2,
    how = "inner",
    on = "key",
    suffixes = ["_d1", "_d2"]
)
print(resultado)
````

Modifica el resultado anterior para que ahora la unión sea de tipo *left*.

````python
resultado = dataset_1.merge(
    dataset_2,
    how = "left",
    on = "key",
    suffixes = ["_d1", "_d2"]
)
print(resultado)
````

<div class="alert alert-info">

Otra forma válida y equivalente de hacer uniones es mediante el uso de la función `pd.merge`. Por ejemplo para este último ejercicio, el código con esta forma sería el siguiente:

```py
resultado = pd.merge(
    dataset_1, dataset_2,
    how = "left",
    on = "key",
    suffixes = ["_d1", "_d2"]
)
print(resultado)
```

</div>

### Ejercicio 2

Considera los siguientes dataframes de ejemplo:

```py
dataset_3 = pd.DataFrame({
    "key_1": ["x", "x", "y", "z"],
    "key_2": [100, -50, 100, -50],
    "col_A": ["a0", "a1", "a2", "a3"],
    "col_B": [True, False, False, True]
})

dataset_4 = pd.DataFrame({
    'key_1': ["x", "y", "y", "z"],
    'key_2': [100, 100, 100, 100],
    'col_C': [2010, 2025, 2021, 2008],
    'col_D': ["d0","d1","d2", "d3"]
})
```

Cárgalos e imprímelos en pantalla.

````python
# Cargar datos
dataset_3 = pd.DataFrame({
    "key_1": ["x", "x", "y", "z"],
    "key_2": [100, -50, 100, -50],
    "col_A": ["a0", "a1", "a2", "a3"],
    "col_B": [True, False, False, True]
})

dataset_4 = pd.DataFrame({
    'key_1': ["x", "y", "y", "z"],
    'key_2': [100, 100, 100, 100],
    'col_C': [2010, 2025, 2021, 2008],
    'col_D': ["d0","d1","d2", "d3"]
})

# Imprimir datos
print(dataset_3)
print()
print(dataset_4)
````

Une los datasets con un *inner join*, usando las llaves key_1 y key_2 a la vez.

````python
resultado = dataset_3.merge(
    dataset_4,
    how = "inner",
    on = ["key_1", "key_2"]
)
print(resultado)
````

Repite la unión anterior pero ahora utiliza un *outer join*.

````python
resultado = dataset_3.merge(
    dataset_4,
    how = "outer",
    on = ["key_1", "key_2"]
)
print(resultado)
````

Ahora usa un *right join*.

````python
resultado = dataset_3.merge(
    dataset_4,
    how = "right",
    on = ["key_1", "key_2"]
)
print(resultado)
````

<div class="alert alert-info">

Existen diferentes tipos de uniones por lo que tu labor como analista o científico de datos será siempre estudiar a detalle el caso en el que estés trabajando y el propósito de la unión que desees realizar. A partir de allí, deberás escoger si usar una unión *left*, *inner*, *outer*, o *right*.

## Aplicación de funciones para enriquecer datos

### Ejercicio 1

Ejecuta el siguiente código para crear un dataset de forma aleatoria.

```py
import pandas as pd
import numpy as np

np.random.seed(123)

# Fechas
fechas = pd.date_range(start="2023-01-01", periods=100, freq="D")

# Datos simulados
ventas = np.random.randint(100, 700, size=100)
gastos_marketing = np.random.randint(50, 200, size=100)
gastos_personal = np.random.randint(80, 300, size=100)

# Crear el DataFrame
df = pd.DataFrame({
    "fecha": fechas,
    "ventas": ventas,
    "gastos_marketing": gastos_marketing,
    "gastos_personal": gastos_personal
})

df.sample(10) 
```

````python
import pandas as pd
import numpy as np

np.random.seed(123)

# Definir fechas cualquiera
fechas = pd.date_range(start="2023-01-01", periods=100, freq="D")

# Simular datos
ventas = np.random.randint(100, 700, size=100)
gastos_marketing = np.random.randint(50, 200, size=100)
gastos_personal = np.random.randint(80, 300, size=100)

# Crear el DataFrame
df = pd.DataFrame({
    "fecha": fechas,
    "ventas": ventas,
    "gastos_marketing": gastos_marketing,
    "gastos_personal": gastos_personal
})

df.sample(10)
````

Crea una nueva columna estado en la que indique si "hay ganancia" cuando las ventas sean mayores a la suma de gastos_marketing y gastos_personal, o "hay pérdida" en caso contrario. Utiliza el método `apply` con el argumento `axis = 1` visto que se requieren varias columnas del dataset en el cálculo deseado.

````python
# Definir función para asignacion
def definir_estado (row):
    if row["ventas"] > row["gastos_marketing"] + row["gastos_personal"]:
        return "hay ganancia"
    else:
        return "hay pérdida"

# Crear nueva columna con metodo apply
df["estado"] = df.apply(definir_estado, axis = 1)
print(df.sample(10))
````

### Ejercicio 2

Crea una nueva columna llamada razon_perdida en la cual se especifique "marketing" si los gastos de este tipo son mayores a los de personal o "personal" en caso contrario, pero solamente cuando el estado sea de pérdida.

````python
def definir_razon (row):
    if row["estado"] == "hay pérdida":
        if row["gastos_marketing"] > row["gastos_personal"]:
            return "marketing"
        else:
            return "personal"
    else:
        return "no aplica"

df["razon"] = df.apply(definir_razon, axis = 1)
df.sample(10)
````

### Ejercicio 3

Si las ventas superan el valor de 500 en un día cualquiera se dice que se ha tenido un "buen día". Crea una columna de tipo booleano que refleje este hecho usando el método `apply`.

````python
def definir_dia_bueno (x):
    if x > 500:
        return True
    else:
        return False

df["es_dia_bueno"] = df["ventas"].apply(definir_dia_bueno)
df.sample(10)
````

<div class="alert alert-info">

Cuando tenemos funciones relativamente sencillas como la del anterior ejercicio, podemos sustituirla por *expresiones lambda* para generar a partir de aquí funciones anónimas y temporales. Por ejemplo, el programa anterior podría transformarse a un código de una sola línea tal que

```py
df["nueva_variable"] = df["ventas"].apply(lambda x: True if x > 500 else False)
```

## Visualizaciones mediante gráficos

### Ejercicio 1

Tomando como referencia el dataset del apartado anterior, crea un gráfico de línea con la evolución de ventas en el tiempo y que tenga varios atributos de forma para hacerlo presentable. Añade además una línea horizontal en el valor de 500 para distinguir los días "buenos". Recuerda importar **matplotlib** antes de crear la visualización.

````python
# Cargar librerias
import matplotlib.pyplot as plt

# Ejecutar grafico
df.plot(
    y = "ventas",
    x = "fecha",
    figsize = [10,4],
    legend = None,
    title = "Evolución de ventas",
    ylabel = "Ventas",
    xlabel = "",
    marker = ".",
    color = "dodgerblue"
)
plt.grid(axis = "y", linestyle = ":")
plt.axhline(y = 500, color = "red", linestyle = "--")
plt.show()
````

### Ejercicio 2

Crea un histograma que muestre y compare la distribución de gastos en marketing y los gastos de personal. Recuerda incluir atributos de forma. Si no sabes cómo crear esta visualización te sugiero consultes en la [documentación](https://matplotlib.org/stable/plot_types/stats/hist_plot.html) de **matplotlib**.

### Ejercicio 3

Crea un gráfico de cajón para tener una idea general de la distribución, dispersión y valores atípicos en las ventas, gastos de marketing y gastos de personal. Recuerda incluir atributos de forma. Si no sabes cómo crear esta visualización te sugiero consultes en la [documentación](https://matplotlib.org/stable/plot_types/stats/boxplot_plot.html) de **matplotlib**.
