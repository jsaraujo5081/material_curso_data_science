### Ejercicio 1

Escribe una función llamada `promedios` que calcule las principales medidas de centralidad de dos números cualquiera y devuelva los resultados a modo de diccionario. Para esto considerar los siguientes estadísticos:

* Media aritmética ($\bar{x}_a$):

$$ \bar{x}_a = \frac{x_1 + x_2}{2} $$

* Media geométrica ($\bar{x}_g$):

$$ \bar{x}_g = \sqrt{|x_1 \times x_2|} $$

* Media armónica ($\bar{x}_h$):

$$ \bar{x}_h = \frac{2}{\frac{1}{x_1} + \frac{1}{x_2}} = \frac{2\times x_1 \times x_2}{x_1 + x_2} $$

````python
def promedios (x_1, x_2):
    med_a = (x_1 + x_2) / 2
    med_g = abs(x_1 * x_2) ** (0.5)
    med_h = (2 * x_1 * x_2) / (x_1 + x_2)
    dicc = {
        "media aritmetica": med_a, 
        "media geometrica": med_g, 
        "media armonica": med_h
    }
    return dicc
````

Comprueba la función creada con dos números cualquiera.

````python
promedios(5.8, 3.7)
````

### Ejercicio 2

Escribe una función llamada `varianza` la cual calcule y devuelva este estadístico ($var$) para dos números cualquiera. Para esto considerar la siguiente definición:

$$ var = \frac{(x_1 - \bar{x}_a)^2 + (x_2 - \bar{x}_a)^2}{2} $$

donde $\bar{x}_a$ es la media aritmética tal como la vimos en el ejercicio anterior.

````python
def varianza (x_1, x_2): 
    med_a = (x_1 + x_2) / 2
    var = ((x_1 - med_a) ** 2 + (x_2 - med_a) ** 2) / 2
    return var
````

Comprueba la función creada con dos números cualquiera. Guarda tu resultado en una variable llamada `var`.

````python
var = varianza(20.3, 12.1)
print(var)
````

Obtén la raíz cuadrada de `var` y guárdala en una variable llamada `desv`. Este resultado se conoce como desviación estándar o típica.

````python
desv = var**(0.5)
print(desv)
````

### Ejercicio 1

Escribe una función llamada `es_bisiesto` que, dado un año `a` como entrada, devuelva si es bisiesto o no. Recuerda que para saber si un año es bisiesto, se debe cumplir con las siguientes condiciones:

* El año debe ser divisible por 4.

* Si es divisible por 100, pero NO es divisible por 400, entonces NO es bisiesto (i.e. 2000 es bisiesto pero 2100 no lo es).

````python
def es_bisiesto (a):
    bisiesto = True
    # Probar primera condicion
    if a % 4 == 0:
        # Probar segunda condicion
        if a % 100 == 0 and a % 400 != 0:
            bisiesto = False
    else:
        bisiesto = False
    return bisiesto
````

Comprueba si los siguientes años son bisiestos: 96, 100, 1492, 1886, 1990, 2000, 4100.

````python
# Crear lista con los años solicitados
periodos = [96, 100, 1492, 1886, 1990, 2000, 4100]

# Generar bucle para comprobar años con funcion
for p in periodos:
    print(f"El año {p} es bisiesto: {es_bisiesto(p)}")
````

### Ejercicio 2

Escribe una función `es_primo` que reciba como entrada un número natural cualquiera y devuelva si el número es primo o compuesto. Vale indicar que un número primo es aquel que solamente es divisible por 1 y por sí mismo, por ejemplo el 7, 11, o 13. Solamente ten cuidado puesto que el número 1 NO es primo.

````python
def es_primo (n):
    if n == 1:
        primo = False
    elif n == 2: 
        primo = True
    else:
        primo = True
        for d in range(2,n):
            if n % d == 0:
                primo = False
                break # El comando break permita detener el bucle sin necesidad de recorrer todas las opciones
    
    return primo
````

Verifica la función creada para los primeros 100 números enteros. Para ello comprueba que existen 25 números primos en este rango. 

````python
cantidad_primos = 0

for n in range(1,101):
    if es_primo(n):
        cantidad_primos = cantidad_primos + 1

print(cantidad_primos)
````

La sucesión de Fibonacci es un conjunto de números naturales muy famoso que deberías conocerlo al menos por cultura general. En esta sucesión se cumple que el número $a$ en la posición $i$ está dado por $a_i = a_{i-1} + a_{i-2}$, con $a_1 = 0$ y $a_2 = 1$. En este sentido, los primeros elementos de esta sucesión son:

$$ 0, 1, 1, 2, 3, 5, 8, 13, ...$$

Utiliza funciones y condicionales para calcular cuál es el número 40 de Fibonacci.


````python
# Definir funcion
def fibonacci (n):
  a1 = 0
  a2 = 1
  if n == 1:
    resultado = a1
  elif n == 2:
    resultado = a2
  else:
    resultado = fibonacci(n-1) + fibonacci(n-2)
  return resultado

# Calcular y mostrar numero 40 de Fibonacci
print(f"Numero 40 de fibonacci: {fibonacci(40):,.0f}")
````

La combinatoria es una operación matemática que permite estimar cuántos grupos posibles de $k$ elementos se pueden crear a partir de un conjunto de tamaño $n$. Esta operación se define por 

$$ \binom{n}{k}  = \frac{n!}{k!(n-k)!} $$

donde $n! = n(n-1)(n-2)...1$ es el factorial de $n$, y $k! = k(k-1)(k-2)...1$ es el factorial de $k$. A partir de esta operación es factible por ejemplo saber que existen 6 posibles grupos de 2 elementos de un conjunto de 4, ya que

$$ \binom{4}{2}  = \frac{4!}{2!(4-2)!} = \frac{4 \times 3 \times 2 \times 1}{2 \times 1 \times 2 \times 1} = 3 \times 2 = 6$$

Una propiedad interesante asociada a la combinatoria es la denominada *Identidad de Pascal*, la cual afirma que:

$$ \binom{n}{k} = \binom{n-1}{k-1} + \binom{n-1}{k}$$

A partir de esta identidad crea una función que tome dos valores $n$ y $k$ cualquiera y devuelva el valor de la combinatoria correspondiente. Para esto considera las siguientes propiedades adicionales de la operación:

* $\binom{n}{0} = 1$

* $\binom{n}{k} = 1$ para $n = k$

* $\binom{n}{k} = 0$ para $k > n$

Verifica tu función comprobando que $\binom{4}{2} = 6$.

````python
# Definir funcion
def combinatoria (n, k):
    if k == 0:
        return 1
    if k == n:
        return 1
    if k > n:
        return 0
    return combinatoria(n-1, k-1) + combinatoria(n-1,k)

# Comprobar funcion
combinatoria(4,2)
````
